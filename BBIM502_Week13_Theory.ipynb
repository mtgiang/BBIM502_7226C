```python
import shutil  # shell utilities - module to handle file operations like copy
from datetime import datetime 

original_file = "file_a.txt"

#Create a timestamp to make the backup filename unique
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

#Create the backup file name using the timestamp
backup_file = f"file_a_backup_{timestamp}.txt"

#Copy the original file to the new backup file
shutil.copy(original_file, backup_file)

# Notify the user that the backup was created
print(f"Backup created: {backup_file}")
```
